```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
mean_GAN1 = np.array([0.61004,0.61104,0.62406,0.62018,0.62222,0.6205,0.618475,0.63354])
sdv_GAN1 = np.array([0.007126458,0.005298906,0.006267567,0.009855638,0.001628987,0.004889172,0.009226694,0.005114528])

mean_GAN2 = np.array([0.62798, 0.63254, 0.63646, 0.63212, 0.63408, 0.63758, 0.639425, 0.64684])
sdv_GAN2 = np.array([0.00486925, 0.003874326, 0.003730201, 0.004067137, 0.005693646, 0.008435022, 0.001287197, 0.006426694])

mean_Noise = np.array([0.6355,0.6401,0.64184,0.64902,0.63194,0.64364,0.64994,0.64848])
sdv_Noise = np.array([0.005033091,0.000616441,0.006431361,0.006330055,0.006902058,0.005094546,0.003113583,0.005601928])

mean_Ori = np.array([0.6145, 0.60344, 0.6241, 0.62706, 0.62756, 0.6282, 0.63776, 0.64106])
sdv_Ori = np.array([0.010557197,0.007954018,0.007591047,0.011070429,0.012129567,0.008916726,0.010092096,0.006413611])

xaxis = np.array([10000, 20000, 30000, 40000, 50000, 60000, 70000, 91324])
```

```python
fig = plt.figure(figsize=(16, 9))
ax = fig.add_subplot(111)

ax.plot(xaxis, mean_GAN1, label = 'DCGAN-Augnmented', linestyle = 'dashed')
ax.fill_between(xaxis, mean_GAN1-sdv_GAN1, mean_GAN1+sdv_GAN1, alpha=.2)

ax.plot(xaxis, mean_GAN2, label = 'cycleGAN-Augnmented', linestyle = 'dashed')
ax.fill_between(xaxis, mean_GAN2-sdv_GAN2, mean_GAN2+sdv_GAN2, alpha=.2)

ax.plot(xaxis, mean_Noise, label = 'Noise-Augnmented',linestyle = 'dashdot')
ax.fill_between(xaxis, mean_Noise-sdv_Noise, mean_Noise+sdv_Noise, alpha=.2)

ax.plot(xaxis, mean_Ori, label = 'Not Augnmented')
ax.fill_between(xaxis, mean_Ori-sdv_Ori, mean_Ori+sdv_Ori, alpha=.1)

ax.set_title('1:1 Augmentation Comparison', fontsize = 15)
ax.legend(loc='best', fontsize = 14)


ax.set_ylabel('AAUC', fontsize = 15)
ax.set_xlabel('Number of Original Images', fontsize = 15)
ax.grid(which = 'major', axis = 'both', alpha = .5)

fig.show()
```
